```python
import 'dart:convert';
import 'package:http/http.dart' as http;

class ApiService {
  // Android emulator: 10.0.2.2 | Real phone: your PC IP, e.g. 192.168.1.5
  static const String baseUrl = 'http://10.0.2.2:8000';

  static Future<dynamic> _get(String path) async {
    final res = await http.get(Uri.parse('$baseUrl$path'));
    if (res.statusCode == 200) return jsonDecode(res.body);
    throw Exception(jsonDecode(res.body)['detail'] ?? 'Request failed');
  }

  static Future<dynamic> _post(String path, Map<String, dynamic> body) async {
    final res = await http.post(
      Uri.parse('$baseUrl$path'),
      headers: {'Content-Type': 'application/json'},
      body: jsonEncode(body),
    );
    if (res.statusCode == 200) return jsonDecode(res.body);
    throw Exception(jsonDecode(res.body)['detail'] ?? 'Request failed');
  }

  // Users
  static Future login(String email, String password) =>
      _post('/login', {'email': email, 'password': password});
  static Future getStudent(int id) => _get('/students/$id');
  static Future getStudentClasses(int id) => _get('/students/$id/classes');

  // Attendance
  static Future scanQr(int studentId, String qrToken) =>
      _post('/attendance/scan', {'student_id': studentId, 'qr_token': qrToken});
  static Future getAttendance(int studentId) => _get('/students/$studentId/attendance');

  // Announcements
  static Future getAnnouncements(int facultyId) => _get('/announcements?faculty_id=$facultyId');

  // Restaurant
  static Future getRestaurants() => _get('/restaurants');
  static Future getMenu(int restaurantId) => _get('/restaurants/$restaurantId/menu');
  static Future createOrder(int studentId, int restaurantId, List<Map<String, int>> items) =>
      _post('/orders', {'student_id': studentId, 'restaurant_id': restaurantId, 'items': items});

  // Trips
  static Future getTrips() => _get('/trips');
  static Future bookTrip(int tripId, int studentId, int seats) =>
      _post('/trips/$tripId/book', {'student_id': studentId, 'seats': seats});

  // Fees
  static Future getFees(int studentId) => _get('/students/$studentId/fees');
  static Future pay(int studentFeeId, double amount) =>
      _post('/payments', {'student_fee_id': studentFeeId, 'amount': amount});
}
```
